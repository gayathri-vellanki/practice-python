```python
#printing number of non-zero elements
l=[1,0,3,0,5,3,0,0,4]
c=0
for i in range(len(l)):
    if l[i]!=0:
        c+=1
print("No of Non-zero elements:",c)
#printing number of zero elements
l=[1,0,3,0,5,3,0,0,4]
c=0
for i in range(len(l)):
    if l[i]==0:
        c+=1
print("No of zero elements:",c)
```

```python
#Printing number of EVEN and ODD elements
l=[2,5,6,33,0,4,56]
even,odd=0,0
for i in range(len(l)):
    if l[i]%2==0:
        even+=1
    else:
        odd+=1
print("EVEN:",even)
print("ODD:",odd)
```

```python
#sum of all even numbers
l=[1,2,3,4,5,6,7]
evSum=0
odSum=0
for i in range(len(l)):
    if l[i]%2==0:
        evSum+=l[i]
    else:
        odSum+=l[i]
if evSum>odSum:
    diff=evSum-odSum
else:
    diff=odSum-evSum
print("Even Sum:",evSum)
print("Odd Sum:",odSum)
print("Difference:",diff)
```

```python
#WITHOUT SORT() ASCENDINGG
h=[5,4,2,3,1]
for i in range(len(h)):
    for j in range(i+1,len(h)):
        if h[i]>=h[j]:
            #h[i],h[j]=h[j],h[i]    without temp
            temp=h[i]
            h[i]=h[j]
            h[j]=temp           
print("Acending Order:",h)
#WITHOUT SORT() DESECNDING
h=[5,4,2,3,1]
for i in range(len(h)):
    for j in range(i+1,len(h)):
        if h[i]<=h[j]:
            #h[i],h[j]=h[j],h[i]    without temp
            temp=h[i]
            h[i]=h[j]
            h[j]=temp         
print("descending order:",h)
```
